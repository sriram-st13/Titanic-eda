# Customer Churn / Sales Trend Analysis
Perform end-to-end analysis on an e-commerce sales dataset to identify top customer segments and revenue trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample e-commerce sales dataset
np.random.seed(42)
dates = pd.date_range(start='2025-01-01', periods=500, freq='D')
df = pd.DataFrame({
    'order_date': np.random.choice(dates, 500),
    'customer_id': np.random.randint(1000, 1050, 500),
    'category': np.random.choice(['Electronics', 'Clothing', 'Home', 'Beauty'], 500),
    'sales_amount': np.random.uniform(20, 500, 500),
    'quantity': np.random.randint(1, 5, 500)
})
df.head()
```

```python
# 1. Feature Engineering & Date Extraction
df['order_date'] = pd.to_datetime(df['order_date'])
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_year'] = df['order_date'].dt.to_period('M')
df.head()
```

```python
# 2. Pivot Tables and Trend Calculations
monthly_sales = df.pivot_table(index='month_year', columns='category', values='sales_amount', aggfunc='sum')
print(monthly_sales.fillna(0))

# Top customer segments by total spending
top_customers = df.groupby('customer_id')['sales_amount'].sum().reset_index().sort_values(by='sales_amount', ascending=False)
print("\nTop 5 Customers by Revenue:")
print(top_customers.head())
```

```python
# 3. Visualizing Trends
plt.figure(figsize=(10, 5))
df.groupby('month_year')['sales_amount'].sum().plot(kind='line', marker='o', color='b')
plt.title('Overall Revenue Trend Over Time')
plt.xlabel('Month-Year')
plt.ylabel('Total Sales Amount')
plt.grid(True)
plt.show()
```
