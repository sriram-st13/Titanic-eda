# Data Visualization with Matplotlib & Seaborn
Create informative charts including histograms, heatmaps, box plots, and correlation matrices from real datasets.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
```

```python
# Load dataset
df = sns.load_dataset('titanic')
df.head()
```

```python
# 1. Bar Charts and Scatter Plots
plt.figure(figsize=(8, 5))
sns.barplot(x='class', y='fare', hue='sex', data=df, palette='muted')
plt.title('Average Fare by Class and Gender')
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(x='age', y='fare', hue='survived', data=df, palette='coolwarm')
plt.title('Age vs Fare Scatter Plot')
plt.show()
```

```python
# 2. Correlation Heatmap Generation
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
